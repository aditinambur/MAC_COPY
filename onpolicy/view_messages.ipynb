```python
import numpy as np
import os
from pathlib import Path
import json
from pprint import pprint
```

```python
# Update this path to point to your .npy file
npy_file_path = "scripts\results\MPE\simple_spread\mappo\check\wandb\offline-run-20260416_203035-o914lgc1\files\messages\episode_000000_step_0000_aggregated_messages.npy"
# Convert to absolute or check if relative path exists
if not os.path.isabs(npy_file_path):
    npy_file_path = os.path.join(os.getcwd(), npy_file_path)

print(f"Loading file: {npy_file_path}")
print(f"File exists: {os.path.exists(npy_file_path)}")
```

```python
# Load the numpy array
messages_array = np.load(npy_file_path)
print(f"Successfully loaded numpy array")
```

```python
# Display metadata
file_stat = os.stat(npy_file_path)

print("=" * 70)
print("FILE METADATA")
print("=" * 70)
print(f"File size: {file_stat.st_size} bytes ({file_stat.st_size / 1024:.2f} KB)")
print(f"File path: {npy_file_path}")
print()

print("=" * 70)
print("ARRAY INFORMATION")
print("=" * 70)
print(f"Array shape: {messages_array.shape}")
print(f"Data type: {messages_array.dtype}")
print(f"Total elements: {messages_array.size}")
print(f"Memory size: {messages_array.nbytes / 1024:.2f} KB")
print(f"Min value: {messages_array.min():.6f}")
print(f"Max value: {messages_array.max():.6f}")
print(f"Mean value: {messages_array.mean():.6f}")
print(f"Std deviation: {messages_array.std():.6f}")
print()
```
